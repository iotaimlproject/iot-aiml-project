# Synthetic OEE Time-Series Dataset Generation

20,000-row dataset for ANN OEE Horizon Prediction Engine

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

print("Libraries loaded successfully")
```

```python
seed_df = pd.read_csv(
    "data/lathe_job_status_data.csv",
    sep=";"
)

print("=== Seed Data ===")
print("Columns:", seed_df.columns.tolist())
print()
print(seed_df.head())
print()
print(f"Shape: {seed_df.shape}")
print()
print("Note: CSV column is 'Avaliability' (typo) -> using 'Availability' in generated data")
```

```python
np.random.seed(42)

n_rows = 20000
start_time = datetime(2026, 1, 1, 0, 0, 0)

timestamps = [start_time + timedelta(minutes=i) for i in range(n_rows)]

rpm = np.zeros(n_rows)
availability = np.zeros(n_rows)
performance = np.zeros(n_rows)
quality = np.zeros(n_rows)
downtime_minutes = np.zeros(n_rows)

rpm[0] = 400.0
availability[0] = 92.0
performance[0] = 78.0
quality[0] = 85.0

dt_memory = 0.0  # lingering downtime effect

for i in range(1, n_rows):
    # --- RPM random walk (smooth) ---
    rpm[i] = rpm[i - 1] + np.random.uniform(-3, 3)
    rpm[i] = np.clip(rpm[i], 300, 500)

    # --- Downtime events (4%) ---
    if np.random.random() < 0.04:
        downtime_minutes[i] = min(np.random.exponential(scale=10), 30)
        dt_memory = downtime_minutes[i]
    else:
        downtime_minutes[i] = 0.0

    if dt_memory > 0 and downtime_minutes[i] == 0:
        dt_memory *= 0.85
        if dt_memory < 0.3:
            dt_memory = 0.0

    dt_active = downtime_minutes[i] + dt_memory

    # --- Performance: random walk with small steps + RPM drift ---
    rpm_drift = (rpm[i] - 400) * 0.01  # tiny drift toward RPM
    dt_penalty_p = dt_active * 1.0
    perf_step = rpm_drift - dt_penalty_p + np.random.normal(0, 0.4)
    performance[i] = performance[i - 1] + perf_step
    performance[i] = np.clip(performance[i], 25, 100)

    # --- Quality: random walk with small steps + degradation + RPM wear ---
    if i % 480 == 0:
        quality[i] = 85.0 + np.random.normal(0, 3)
    else:
        q_drift = -0.005 - (rpm[i] - 300) * 0.0003  # slow degradation
        q_dt = dt_active * 0.5
        q_noise = np.random.normal(0, 0.5)
        quality[i] = quality[i - 1] + q_drift - q_dt + q_noise
    quality[i] = np.clip(quality[i], 35, 100)

    # --- Availability: random walk with small steps ---
    avail_dt = dt_active * 2.0
    avail_recovery = 0.02 if dt_active == 0 and availability[i-1] < 90 else 0.0
    avail_step = avail_recovery - avail_dt + np.random.normal(0, 0.3)
    availability[i] = availability[i - 1] + avail_step
    availability[i] = np.clip(availability[i], 40, 100)

current_oee = (availability * performance * quality) / 10000.0

df = pd.DataFrame({
    'timestamp': timestamps,
    'rpm': rpm.round(1),
    'availability': availability.round(2),
    'performance': performance.round(2),
    'quality': quality.round(2),
    'current_oee': current_oee.round(2),
    'downtime_minutes': downtime_minutes.round(2)
})

print(f"Generated {len(df)} rows")

print("\n=== Correlation: current state vs target_oee_1h ===")
temp = df.copy()
temp['target_oee_1h'] = temp['current_oee'].shift(-60)
temp = temp.dropna()
for col in ['current_oee', 'rpm', 'availability', 'performance', 'quality', 'downtime_minutes']:
    c = temp[col].corr(temp['target_oee_1h'])
    print(f"  {col:20s} → target_oee_1h: {c:.3f}")

print("\n=== State correlations ===")
corr_cols = ['rpm', 'availability', 'performance', 'quality', 'current_oee', 'downtime_minutes']
print(df[corr_cols].corr().round(3))

print("\n=== Distributions ===")
for col in ['availability', 'performance', 'quality', 'current_oee']:
    print(f"  {col:15s} min={df[col].min():6.1f}  max={df[col].max():6.1f}  mean={df[col].mean():6.1f}  std={df[col].std():6.1f}")

print(f"\n  Downtime events: {(df['downtime_minutes']>0).sum()}/{len(df)} = {(df['downtime_minutes']>0).mean()*100:.1f}%")
print(f"  Mean OEE when downtime>0: {df[df['downtime_minutes']>0]['current_oee'].mean():.1f}")
print(f"  Mean OEE when downtime=0: {df[df['downtime_minutes']==0]['current_oee'].mean():.1f}")
```

```python
df['target_oee_30m'] = df['current_oee'].shift(-30)
df['target_oee_1h'] = df['current_oee'].shift(-60)
df['target_oee_2h'] = df['current_oee'].shift(-120)
df['target_oee_6h'] = df['current_oee'].shift(-360)
df['target_oee_8h'] = df['current_oee'].shift(-480)

before = len(df)
df = df.dropna().reset_index(drop=True)
after = len(df)
print(f"Rows before dropna: {before}")
print(f"Rows after dropna: {after}")
print(f"Rows removed: {before - after}")
```

```python
df.to_csv("data/syn_timeseries_oee.csv", index=False)
print(f"Saved: {df.shape}")
```

```python
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

plot_df = df.head(500)

axes[0].plot(plot_df['timestamp'], plot_df['rpm'], color='blue', linewidth=0.8)
axes[0].set_ylabel('RPM')
axes[0].set_title('RPM over Time (first 500 rows)')
axes[0].grid(True, alpha=0.3)

axes[1].plot(plot_df['timestamp'], plot_df['availability'],
             label='Availability', linewidth=0.8)
axes[1].plot(plot_df['timestamp'], plot_df['performance'],
             label='Performance', linewidth=0.8)
axes[1].plot(plot_df['timestamp'], plot_df['quality'],
             label='Quality', linewidth=0.8)
axes[1].set_ylabel('%')
axes[1].set_title('Availability / Performance / Quality (first 500 rows)')
axes[1].legend(fontsize=8)
axes[1].grid(True, alpha=0.3)

axes[2].plot(plot_df['timestamp'], plot_df['current_oee'],
             label='Current OEE', color='green', linewidth=0.8)
axes[2].fill_between(
    plot_df['timestamp'], 0,
    plot_df['downtime_minutes'] * 3,
    alpha=0.3, color='red', label='Downtime (scaled x3)'
)
axes[2].set_ylabel('OEE / Downtime')
axes[2].set_title('OEE and Downtime Events (first 500 rows)')
axes[2].legend(fontsize=8)
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
